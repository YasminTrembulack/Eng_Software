### Métricas de classificação
* y = categórico/nominal


```python
# Bibliotecas
from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    accuracy_score,
    recall_score,
    f1_score,
)
```

```python
# Dados de exemplo
y       = [0, 1, 0, 1, 0, 1] # Real
y_pred  = [0, 0, 0, 1, 0, 1] # Previsto
```

```python
print("Métricas de classificação:")

# Acurácia: proporção de acertos.
accuracy = accuracy_score(y, y_pred)
print(f" - Acurácia: {(accuracy * 100):.2f}%")

# Precisão: dos que foram classificados como positivo, quantos realmente são.
precision = precision_score(y, y_pred)
print(f" - Precisão: {(precision * 100):.2f}%")

# Revocação: dos possitivos reais, quantos o modelo encontrou
recall = recall_score(y, y_pred)
print(f" - Revocação: {(recall * 100):.2f}%")

#* F1 Score: equilibrío entre precisão e revocação
f1 = f1_score(y, y_pred)
print(f" - F1 Score: {(f1 * 100):.2f}%")
```

```python
# Síntese
from sklearn.metrics import classification_report

print(classification_report(y, y_pred))
```

```python
# Matriz de confusão
# Mostra os acertos e erros por classe
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(
    confusion_matrix(y, y_pred),
    annot=True,
    fmt='d',
    cmap='YlGnBu'    
)

# Diagonal (esquerda) principal é a que mais importa

plt.title('Matriz de confusão')
plt.xlabel('Previsto')
plt.ylabel('Real')
plt.show()
```

### Métricas de regressão
* y = numérico

```python
# Biblioteca
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
# Dados de exemplo
y =      [3.0, -0.5, 2.0, 7.0]
y_pred = [2.5, 0.0, 2.0, 8.0]  
```

```python
print("Métricas de regressão:")

# Mean Absolute Error (MAE): erro médio absoluto -> fácil de interpretar.
mae = mean_absolute_error(y, y_pred)
print(f" - MAE: {mae:.2f}")

# Mean Squared Error (MSE): penaliza erros grandes (obj -> minimizar)
mse = mean_squared_error(y, y_pred)
print(f" - MSE: {mse:.2f}")

# R²: capacidade de explicação (obj -> maximizar)
r2 = r2_score(y, y_pred)
print(f" - R²: {r2:.2f}")
```
