```python
# Importando bibliotecas
import cv2
import matplotlib.pyplot as plt
```

```python
# 1️⃣ Carregar uma imagem colorida
# Substitua 'sua_imagem.jpg' pelo caminho da sua imagem
img_color = cv2.imread('./cat_felix.jpg')  # Carrega a imagem
img_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)  # Converte BGR -> RGB para exibição correta
```

```python
# 2️⃣ Mostrar a imagem colorida
plt.figure(figsize=(6,6))
plt.imshow(img_color)
plt.title("Imagem Colorida")
plt.axis('off')
plt.show()
```

```python
# 3️⃣ Mostrar os dados da imagem (array)
print("Array da imagem colorida:\n", img_color)
```

```python
# 4️⃣ Mostrar as dimensões
print("Dimensões da imagem colorida:", img_color.shape)  # (altura, largura, canais)
```

```python
# 5️⃣ Converter para escala de cinza
img_gray = cv2.cvtColor(img_color, cv2.COLOR_RGB2GRAY)
```

```python
# 6️⃣ Mostrar a imagem em escala de cinza
plt.figure(figsize=(6,6))
plt.imshow(img_gray, cmap='gray')
plt.title("Imagem em Escala de Cinza")
plt.axis('off')
plt.show()
```

```python
# 7️⃣ Mostrar as dimensões da imagem em cinza
print("Dimensões da imagem em cinza:", img_gray.shape)  # (altura, largura)
```

```python
# 8️⃣ Incrementar o contraste na imagem em cinza
# Vamos usar a técnica de equalização de histograma
img_gray_eq = cv2.equalizeHist(img_gray)
```

```python
# 9️⃣ Mostrar os histogramas antes e depois
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(img_gray.ravel(), bins=256, range=[0,256], color='gray')
plt.title("Histograma Antes do Contraste")

plt.subplot(1,2,2)
plt.hist(img_gray_eq.ravel(), bins=256, range=[0,256], color='gray')
plt.title("Histograma Após o Contraste")

plt.show()
```

```python
# 🔟 Mostrar as duas imagens lado a lado
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Cinza")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img_gray_eq, cmap='gray')
plt.title("Contraste Incrementado")
plt.axis('off')

plt.show()
```

```python
# Importando bibliotecas
import cv2
import matplotlib.pyplot as plt

# 1️⃣ Carregar uma imagem colorida
# Substitua 'sua_imagem.jpg' pelo caminho da sua imagem
img_color = cv2.imread('./cat_felix.jpg')  # Carrega a imagem
img_color = cv2.cvtColor(img_color, cv2.COLOR_BGR2RGB)  # Converte BGR -> RGB para exibição correta

# 2️⃣ Mostrar a imagem colorida
plt.figure(figsize=(6,6))
plt.imshow(img_color)
plt.title("Imagem Colorida")
plt.axis('off')
plt.show()

# 3️⃣ Mostrar os dados da imagem (array)
print("Array da imagem colorida:\n", img_color)

# 4️⃣ Mostrar as dimensões
print("Dimensões da imagem colorida:", img_color.shape)  # (altura, largura, canais)

# 5️⃣ Converter para escala de cinza
img_gray = cv2.cvtColor(img_color, cv2.COLOR_RGB2GRAY)

# 6️⃣ Mostrar a imagem em escala de cinza
plt.figure(figsize=(6,6))
plt.imshow(img_gray, cmap='gray')
plt.title("Imagem em Escala de Cinza")
plt.axis('off')
plt.show()

# 7️⃣ Mostrar as dimensões da imagem em cinza
print("Dimensões da imagem em cinza:", img_gray.shape)  # (altura, largura)

# 8️⃣ Incrementar o contraste na imagem em cinza
# Vamos usar a técnica de equalização de histograma
img_gray_eq = cv2.equalizeHist(img_gray)

# 9️⃣ Mostrar os histogramas antes e depois
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(img_gray.ravel(), bins=256, range=[0,256], color='gray')
plt.title("Histograma Antes do Contraste")

plt.subplot(1,2,2)
plt.hist(img_gray_eq.ravel(), bins=256, range=[0,256], color='gray')
plt.title("Histograma Após o Contraste")

plt.show()

# 🔟 Mostrar as duas imagens lado a lado
plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
plt.imshow(img_gray, cmap='gray')
plt.title("Original Cinza")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(img_gray_eq, cmap='gray')
plt.title("Contraste Incrementado")
plt.axis('off')

plt.show()
```
