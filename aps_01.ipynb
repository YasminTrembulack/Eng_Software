```python
from collections import deque
from typing import Dict
import heapq

grafo = {}
```

```python
def bfs(grafo: Dict, inicio: str):
    fila = deque([inicio])
    visitado = set([inicio])
    ordem = []
    
    while fila:
        atual = fila.popleft()
        ordem.append(atual)

        for v in grafo.get(atual):
            if v not in visitado:
                fila.append(v)
                visitado.add(v)
    return ordem
```

```python
def a_star(grafo, heuristica, inicio, destino):
    fila, visitados  = [(heuristica[inicio], 0, inicio, [inicio])], set()
    while fila:
        _, g, atual, caminho = heapq.heappop(fila)
        if atual in visitados:
            continue
        if atual == destino:
            return caminho, g + heuristica[atual]
        visitados.add(atual)
        for vizinho, custo in grafo[atual].items():
            if vizinho not in visitados:
                heapq.heappush(fila, 
                    (g + custo + heuristica[vizinho],
                    g + custo,
                    vizinho,
                    caminho + [vizinho])
                )     
    return None, float('inf')  
```

```python
def show_result():
    largura = max(len(e) for e in grafo)
    for i in grafo.keys():
        result = bfs(grafo, i)
        print(f"Ordem BFS - {i.upper().ljust(largura)}:  " + str(result))
```
