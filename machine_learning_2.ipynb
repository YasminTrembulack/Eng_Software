```python
import numpy as np
```

```python
# Funções auxiliares
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivada(x):
    return x * (1 - x)
```

```python
# Dados entrada e saída
entradas = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1],
])

saidas = np.array([[0], [1], [1], [0]])
```

```python
# Inicie com pesos aleatórios
np.random.seed(42)
pesos_0 = 2 * np.random.rand(2, 4) -1 # 2 entradas, 4 neurônios
pesos_1 = 2 * np.random.rand(4, 1) -1 # neurônios -> 1 saida
```

```python
# Treinamentos
for epoca in range(1000):
    camada_0 = entradas # Entrada da rede
    camada_1 = sigmoid(np.dot(camada_0, pesos_0)) # Saída camada
    camada_2 = sigmoid(np.dot(camada_1, pesos_1)) # Saída final
    
    # Calculo do erro
    erro = saidas - camada_2
    
    # Imprimir as épocas
    if epoca % 100 == 0:
        print(f'Época {epoca} - Erro médio: {np.mean(np.abs(erro)):.4f}')
```

```python
# Ajusar os pesos (backpropagation)
delta_saida = erro * sigmoid_derivada(camada_2)
delta_oculta = delta_saida.dot(pesos_1.T) * sigmoid_derivada(camada_1)

# Atualizar os pesos
pesos_0 = camada_0.T.dot(delta_oculta)
pesos_1 = camada_1.T.dot(delta_saida)
```
