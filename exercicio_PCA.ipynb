```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from PIL import Image
```

```python
# ==========================
# 1️⃣ Importar e exibir a imagem
# ==========================
# Substitua pelo nome da sua imagem
img_path = "./cat_felix.jpg"  
img = Image.open(img_path)

plt.figure(figsize=(4, 4))
plt.imshow(img)
plt.title("Imagem Original (Colorida)")
plt.axis("off")
plt.show()
```

```python
# ==========================
# 2️⃣ Converter para escala de cinza
# ==========================
gray_img = img.convert("L")
plt.figure(figsize=(4, 4))
plt.imshow(gray_img, cmap="gray")
plt.title("Imagem em Escala de Cinza")
plt.axis("off")
plt.show()
```

```python
# ==========================
# 3️⃣ Mostrar dimensões (pixels)
# ==========================
img_array = np.array(gray_img)
print(f"Dimensões da imagem: {img_array.shape}")
```

```python
# ==========================
# 4️⃣ Converter para formato padrão (matriz 2D)
# ==========================
# Cada linha será um pixel (linha da imagem)
X = img_array / 255.0  # Normaliza os valores (0–1)

print(f"Formato para PCA: {X.shape}")
```

```python
# ==========================
# 5️⃣ Aplicar PCA
# ==========================
pca = PCA()
pca.fit(X)
```

```python
# ==========================
# 6️⃣ Verificar variância explicada cumulativa
# ==========================
var_cum = np.cumsum(pca.explained_variance_ratio_)

plt.figure()
plt.plot(var_cum * 100)
plt.xlabel('Número de Componentes Principais')
plt.ylabel('Variância Explicada Acumulada (%)')
plt.title('Variância Explicada pela PCA')
plt.grid(True)
plt.show()
```

```python
# ==========================
# 7️⃣ Redução para 95% e 99% da variância
# ==========================
pca_95 = PCA(0.95)
X_pca_95 = pca_95.fit_transform(X)
X_reconstructed_95 = pca_95.inverse_transform(X_pca_95)

pca_99 = PCA(0.99)
X_pca_99 = pca_99.fit_transform(X)
X_reconstructed_99 = pca_99.inverse_transform(X_pca_99)
```

```python
# ==========================
# 8️⃣ Mostrar imagens reconstruídas
# ==========================
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].imshow(X, cmap="gray")
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(X_reconstructed_95, cmap="gray")
ax[1].set_title("Reconstruída (95%)")
ax[1].axis("off")

ax[2].imshow(X_reconstructed_99, cmap="gray")
ax[2].set_title("Reconstruída (99%)")
ax[2].axis("off")

plt.show()
```

```python
# ==========================
# 9️⃣ Mostrar quantidade de características
# ==========================
print(f"Número original de características: {X.shape[1]}")
print(f"Características mantidas (95%): {pca_95.n_components_}")
print(f"Características mantidas (99%): {pca_99.n_components_}")
```

```python
# ==========================
# 🔟 Gráfico das projeções e autovetores
# ==========================
plt.figure(figsize=(6, 6))
plt.scatter(X_pca_95[:, 0], X_pca_95[:, 1], s=2, alpha=0.5, label="Projeções (pixels)")
plt.quiver(np.zeros(pca_95.components_.shape[0]),
           np.zeros(pca_95.components_.shape[0]),
           pca_95.components_[:, 0],
           pca_95.components_[:, 1],
           angles='xy', scale_units='xy', scale=1, color='r', label="Autovetores (PCs)")
plt.title("Projeções e Autovetores (PCA 95%)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from PIL import Image

# ==========================
# 1️⃣ Importar e exibir a imagem
# ==========================
# Substitua pelo nome da sua imagem
img_path = "./tiger-jpg.jpg"  
img = Image.open(img_path)

plt.figure(figsize=(4, 4))
plt.imshow(img)
plt.title("Imagem Original (Colorida)")
plt.axis("off")
plt.show()

# ==========================
# 2️⃣ Converter para escala de cinza
# ==========================
gray_img = img.convert("L")
plt.figure(figsize=(4, 4))
plt.imshow(gray_img, cmap="gray")
plt.title("Imagem em Escala de Cinza")
plt.axis("off")
plt.show()

# ==========================
# 3️⃣ Mostrar dimensões (pixels)
# ==========================
img_array = np.array(gray_img)
print(f"Dimensões da imagem: {img_array.shape}")

# ==========================
# 4️⃣ Converter para formato padrão (matriz 2D)
# ==========================
# Cada linha será um pixel (linha da imagem)
X = img_array / 255.0  # Normaliza os valores (0–1)

print(f"Formato para PCA: {X.shape}")

# ==========================
# 5️⃣ Aplicar PCA
# ==========================
pca = PCA()
pca.fit(X)

# ==========================
# 6️⃣ Verificar variância explicada cumulativa
# ==========================
var_cum = np.cumsum(pca.explained_variance_ratio_)

plt.figure()
plt.plot(var_cum * 100)
plt.xlabel('Número de Componentes Principais')
plt.ylabel('Variância Explicada Acumulada (%)')
plt.title('Variância Explicada pela PCA')
plt.grid(True)
plt.show()

# ==========================
# 7️⃣ Redução para 95% e 99% da variância
# ==========================
pca_95 = PCA(0.95)
X_pca_95 = pca_95.fit_transform(X)
X_reconstructed_95 = pca_95.inverse_transform(X_pca_95)

pca_99 = PCA(0.99)
X_pca_99 = pca_99.fit_transform(X)
X_reconstructed_99 = pca_99.inverse_transform(X_pca_99)

# ==========================
# 8️⃣ Mostrar imagens reconstruídas
# ==========================
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
ax[0].imshow(X, cmap="gray")
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(X_reconstructed_95, cmap="gray")
ax[1].set_title("Reconstruída (95%)")
ax[1].axis("off")

ax[2].imshow(X_reconstructed_99, cmap="gray")
ax[2].set_title("Reconstruída (99%)")
ax[2].axis("off")

plt.show()

# ==========================
# 9️⃣ Mostrar quantidade de características
# ==========================
print(f"Número original de características: {X.shape[1]}")
print(f"Características mantidas (95%): {pca_95.n_components_}")
print(f"Características mantidas (99%): {pca_99.n_components_}")

# ==========================
# 🔟 Gráfico das projeções e autovetores
# ==========================
plt.figure(figsize=(6, 6))
plt.scatter(X_pca_95[:, 0], X_pca_95[:, 1], s=2, alpha=0.5, label="Projeções (pixels)")
plt.quiver(np.zeros(pca_95.components_.shape[0]),
           np.zeros(pca_95.components_.shape[0]),
           pca_95.components_[:, 0],
           pca_95.components_[:, 1],
           angles='xy', scale_units='xy', scale=1, color='r', label="Autovetores (PCs)")
plt.title("Projeções e Autovetores (PCA 95%)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.legend()
plt.grid(True)
plt.show()
```
