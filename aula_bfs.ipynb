```python
from collections import deque
from typing import Dict
```

```python
grafo = {
    'A':['B', 'C'],
    'B':['A', 'B', 'E'],
    'C':['A', 'D'],
    'D':['B', 'C', 'E'],
    'E':['B', 'D']
}
```

```python
def bfs(grafo: Dict, inicio: str):
    fila = deque([inicio])
    visitado = set([inicio])
    ordem = [] # ordem de visitação
    
    while fila:
        atual = fila.popleft() # remove o primeiro elemento da fila
        ordem.append(atual) # registrar visita
        # Adicionar vizinhos não visitados a fila
        for v in grafo.get(atual):
            if v not in visitado:
                fila.append(v)
                visitado.add(v)
    return ordem
```

```python
ordem = bfs(grafo, 'E')
print(ordem)
```

```python
def minha_busca_em_largura(grafo: Dict, inicio: str, fim: str):
    fila = deque([inicio])
    visitado = set([inicio])
    ordem = [] # ordem de visitação
    
    while True:
        atual = fila.popleft() # remove o primeiro elemento da fila
        ordem.append(atual) # registrar visita
        if atual == fim:
            return ordem
        # Adicionar vizinhos não visitados a fila
        for v in grafo.get(atual):
            if v not in visitado:
                fila.append(v)
                visitado.add(v)
```

```python
ordem_2 = minha_busca_em_largura(grafo, 'E', 'A')
print(ordem_2)
```

```python
cidades = {
    'Centro': ['Afonso Pena', 'Costeira', 'Boneca do Iguaçu'],
    'Afonso Pena': ['Centro', 'São Domingos', 'Guatupê'],
    'São Domingos': ['Afonso Pena', 'Braga'],
    'Costeira': ['Centro', 'Guatupê', 'Iná'],
    'Guatupê': ['Afonso Pena', 'Costeira', 'Campo Largo da Roseira'],
    'Braga': ['São Domingos'],
    'Iná': ['Costeira'],
    'Boneca do Iguaçu': ['Centro'],
    'Campo Largo da Roseira': ['Guatupê']
}
```

```python
inicio = ['Centro', 'Iná', 'Campo Largo da Roseira']

for i in inicio:
    caminho = bfs(cidades, i)
    print(i + ": " + str(caminho))
```
