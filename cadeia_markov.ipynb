```python
import numpy as np
```

```python
nomes_estados = [ 'Ensolarado', 'Nublado', 'Chuvoso']
```

```python
# Matriz de transição
matriz = np.array([
    [0.7, 0.2, 0.1],
    [0.3, 0.4, 0.3],
    [0.2, 0.3, 0.5]
])
matriz.shape
```

```python
estado_inicial = np.array([0.9, 0.1, 0.0])
```

```python
def cadeia_markov(matriz, estado_inicial, periodo):
    estados = [estado_inicial] # Armazena estados ao longo do tempo
    if isinstance(periodo, str) or isinstance(periodo, list):
        periodo = len(periodo)
        
    for _ in range(periodo):
        # Multiplica a matriz de transição pelo estado atual (vetor)
        novo_estado = np.dot(estados[-1], matriz)
        estados.append(novo_estado) # Adiciona novo estado a lista
    # Converte para array para facilitar a manipulação
    estados = np.array(estados).round(3) 
    return estados
```

```python
cadeia_markov(matriz, estado_inicial, nomes_estados)
```

```python
# Informar numero de periodos
# periodos = int(input("Por favor, informe o número de periodos: "))

simulacao = cadeia_markov(matriz, estado_inicial, "periodos")

#Exibir resultados
def print_simulacao(simulacao):
    print("Estados simulados ao longo dos dias")
    for i, estado in enumerate(simulacao):
        print(f"\nDia {i}: ")
        for j, probabilidade in enumerate(estado):
            print(f" {nomes_estados[j]}: {probabilidade}")
        
        
```

Uma empresa quer entender o comportamento de seus clientes ao navegar em
seu site e modela o processo como uma cadeia de Markov com três estados:
NAVEGANDO, NO CARRINHO e COMPRANDO. Um cliente pode permanecer no mesmo estado
ou transitar para outro com certas probabilidades.

Se um cliente está navegando, há 60% de chance de continuar navegando,
30% de chance de adicionar algo ao carrinho e uma probabilidade desconhecida de
concluir a compra diretamente. 

Se o cliente já está no carrinho, há uma probabilidade desconhecida de voltar a navegar,
50% de chance de permanecer no carrinho e 40% de chance de finalizar a compra. 

Já um cliente que está comprando tem 20% de chance de voltar a navegar, uma probabilidade
desconhecida de retornar ao carrinho e 70% de chance de concluir a compra.

Inicialmente, a empresa estima que 50% dos clientes começam apenas
navegando, enquanto 30% já possuem itens no carrinho, e os 20% restantes estão no
processo de compra.

```python

nomes_estados = [ 'Navegando', 'No Carrinho', 'Comprando']
estado_inicial = np.array([0.5, 0.3, 0.2])
matriz = np.array([
    [0.6, 0.3, 0.1],
    [0.1, 0.5, 0.4],
    [0.2, 0.1, 0.7]
])
```

```python
simulacao = cadeia_markov(matriz, estado_inicial, 10)
print_simulacao(simulacao)
```
