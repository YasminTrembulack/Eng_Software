```python
# Bibliotecas
from sklearn.datasets import (
    fetch_california_housing,
    load_breast_cancer,
    load_diabetes,
    load_digits,
)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.neural_network import (
    MLPClassifier,
    MLPRegressor
)
from sklearn.metrics import (
    classification_report,
    mean_absolute_error,
    mean_squared_error,
    accuracy_score,
    r2_score,
)
```

# 1 QUESTÃO

```python
# Carregar dados
data = load_breast_cancer()
X, y = data.data, data.target
```

```python
# Dividir dados
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
# Normalização
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Função para avaliar e imprimir resultados
def avaliar_modelo(modelo, nome):
    modelo.fit(X_train_scaled, y_train)
    y_pred = modelo.predict(X_test_scaled)
    print(f"\n------------- {nome} -------------")
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print(classification_report(y_test, y_pred, target_names=data.target_names))
    
```

```python
# Cenário 1: Perceptron
modelo_1 = Perceptron(max_iter=1000, random_state=42)
avaliar_modelo(modelo_1, "Perceptron (baseline)", )
```

```python
# Cenário 2: Perceptron
modelo_1 = Perceptron(max_iter=200, random_state=42)
avaliar_modelo(modelo_1, "Perceptron (baseline)")
```

```python
# Cenário 3: MLP com 1 camada oculta
modelo_3 = MLPClassifier(
    hidden_layer_sizes=(30,), activation='relu', max_iter=150, random_state=42
)
avaliar_modelo(modelo_3, "MLP (30,) - relu")
```

```python
# Cenário 4: MLP com 2 camadas ocultas
modelo_4 = MLPClassifier(
    hidden_layer_sizes=(50, 30), activation='relu', max_iter=1000, random_state=42
)
avaliar_modelo(modelo_4, "MLP (50, 30) - relu")
```

```python
# Cenário 5: MLP com ativação tanh
modelo_5 = MLPClassifier(
    hidden_layer_sizes=(30,), activation='tanh', max_iter=1000, random_state=42
)
avaliar_modelo(modelo_5, "MLP (30,) - tanh")
```

```python
# Cenário 6: MLP sem normalização (para comparação)
modelo_6 = MLPClassifier(
    hidden_layer_sizes=(30,), activation='relu', max_iter=1000, random_state=42
)
modelo_6.fit(X_train, y_train)
y_pred_6 = modelo_6.predict(X_test)
print("\n== MLP (30,) - relu - sem normalização ==")
print("Accuracy:", accuracy_score(y_test, y_pred_6))
print(classification_report(y_test, y_pred_6, target_names=data.target_names))
```

# 2 QUESTÃO

```python
# Carregar dados
digits = load_digits()
X, y = digits.data, digits.target
```

```python
# Dividir dados
X_train, X_test, y_train, y_test = train_test_split(
    X, y, random_state=42, test_size=0.3, stratify=y
)
```

```python
# Normalizar dados
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
def avaliar_modelo(modelo, nome, X_tr, X_te, y_tr, y_te):
    modelo.fit(X_tr, y_tr)
    y_pred = modelo.predict(X_te)
    acc = accuracy_score(y_te, y_pred)
    print(f"\n===== {nome} =====")
    print(f"Accuracy: {acc:.4f}")
    print(classification_report(y_te, y_pred))
```

```python
# Cenário 1: Perceptron padrão (sem normalização)
model_1 = Perceptron(max_iter=1000, tol=1e-3, random_state=42)
avaliar_modelo(model_1, "Perceptron - sem normalização", X_train, X_test, y_train, y_test)
```

```python
# Cenário 2: Perceptron com normalização
model_2 = Perceptron(max_iter=1000, tol=1e-3, random_state=42)
avaliar_modelo(model_2, "Perceptron - com normalização", X_train_scaled, X_test_scaled, y_train, y_test)
```

```python
# Cenário 3: MLP 1 camada oculta (30 neurônios), relu, normalizado
model_3 = MLPClassifier(hidden_layer_sizes=(30,), activation='relu', max_iter=200, random_state=42)
avaliar_modelo(model_3, "MLP (30,) - relu - normalizado", X_train_scaled, X_test_scaled, y_train, y_test)
```

```python
# Cenário 4: MLP 2 camadas ocultas (50, 30), relu, normalizado
model_4 = MLPClassifier(hidden_layer_sizes=(50, 30), activation='relu', max_iter=1000, random_state=42)
avaliar_modelo(model_4, "MLP (50,30) - relu - normalizado", X_train_scaled, X_test_scaled, y_train, y_test)
```

```python
# Cenário 5: MLP 1 camada oculta (50), tanh, normalizado
model_5 = MLPClassifier(hidden_layer_sizes=(50,), activation='tanh', max_iter=1000, random_state=42)
avaliar_modelo(model_5, "MLP (50,) - tanh - normalizado", X_train_scaled, X_test_scaled, y_train, y_test)
```

# 3 QUESTÃO

```python
# Carregamento dos dados
data = load_diabetes()
X, y = data.data, data.target
```

```python
# Divisão em treino e teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Normalização
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Função para avaliar modelo
def avaliar_modelo_2(modelo, nome, X_tr, X_te):
    modelo.fit(X_tr, y_train)
    y_pred = modelo.predict(X_te)
    print(f"\n====== {nome} ======")
    print("MSE:", mean_squared_error(y_test, y_pred))
    print("MAE:", mean_absolute_error(y_test, y_pred))
    print("R²:", r2_score(y_test, y_pred))
```

```python
# Cenário 1: 1 camada oculta (30), relu, 300 épocas
modelo_1 = MLPRegressor(hidden_layer_sizes=(30,), activation='relu', max_iter=300, random_state=42)
avaliar_modelo_2(modelo_1, "MLP (30,) - relu - 300 épocas", X_train_scaled, X_test_scaled)
```

```python
# Cenário 2: 2 camadas ocultas (50, 30), relu, 500 épocas
modelo_2 = MLPRegressor(hidden_layer_sizes=(50, 30), activation='relu', max_iter=500, random_state=42)
avaliar_modelo_2(modelo_2, "MLP (50, 30) - relu - 500 épocas", X_train_scaled, X_test_scaled)
```

```python
# Cenário 3: 1 camada oculta (100), tanh, 1000 épocas
modelo_3 = MLPRegressor(hidden_layer_sizes=(100,), activation='tanh', max_iter=1000, random_state=42)
avaliar_modelo_2(modelo_3, "MLP (100,) - tanh - 1000 épocas", X_train_scaled, X_test_scaled)
```

```python
# Cenário 4: 2 camadas (100, 50), tanh, 1000 épocas
modelo_4 = MLPRegressor(hidden_layer_sizes=(100, 50), activation='tanh', max_iter=1000, random_state=42)
avaliar_modelo_2(modelo_4, "MLP (100, 50) - tanh - 1000 épocas", X_train_scaled, X_test_scaled)
```

```python
# Cenário 5: 1 camada oculta (50), logistic, 500 épocas
modelo_5 = MLPRegressor(hidden_layer_sizes=(50,), activation='logistic', max_iter=500, random_state=42)
avaliar_modelo_2(modelo_5, "MLP (50,) - logistic - 500 épocas", X_train_scaled, X_test_scaled)
```

```python
# Cenário 6: 1 camada oculta (30), relu, 1000 épocas - SEM normalização
modelo_6 = MLPRegressor(hidden_layer_sizes=(30,), activation='relu', max_iter=1000, random_state=42)
avaliar_modelo_2(modelo_6, "MLP (30,) - relu - 1000 épocas - SEM normalização", X_train, X_test)
```
