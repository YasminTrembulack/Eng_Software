```python
# Bibliotecas
from sklearn.datasets import (
    fetch_california_housing,
    load_breast_cancer,
    load_diabetes,
    load_digits,
)
from sklearn.neural_network import MLPClassifier, MLPRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.metrics import (
    classification_report,
    mean_absolute_error,
    mean_squared_error,
    accuracy_score,
    r2_score,
)
```

```python
# Função para avaliar e imprimir resultados
def avaliar_modelo_classificacao(modelo, X_test, y_test, nome_modelo):
    y_pred = modelo.predict(X_test)
    print(f"\n------------- Modelo: {nome_modelo} -------------")
    print(classification_report(y_test, y_pred))
    print("Acurácia:", accuracy_score(y_test, y_pred))

def avaliar_modelo_regressao(modelo, X_test, y_test, nome_modelo):
    y_pred = modelo.predict(X_test)
    print(f"\n------------- Modelo: {nome_modelo} -------------")
    print("MAE:", mean_absolute_error(y_test, y_pred))
    print("MSE:", mean_squared_error(y_test, y_pred))
    print("R2:", r2_score(y_test, y_pred))
```

## Questão 1 - Breast Cancer (Classificação)

```python
# Carregar dados
data_bc = load_breast_cancer()
X_bc, y_bc = data_bc.data, data_bc.target

# Dividir e treinar dados
X_train_bc, X_test_bc, y_train_bc, y_test_bc = train_test_split(
    X_bc, y_bc, test_size=0.3, random_state=42
)

# Normalização
scaler_bc = StandardScaler()
X_train_bc_scaled = scaler_bc.fit_transform(X_train_bc)
X_test_bc_scaled = scaler_bc.transform(X_test_bc)
```

### Cenários

```python
modelo_1 = Perceptron(max_iter=1000, random_state=42)
modelo_1.fit(X_train_bc_scaled, y_train_bc)
avaliar_modelo_classificacao(modelo_1, X_test_bc_scaled, y_test_bc, "Perceptron")
```

```python
modelo_2 = Perceptron(max_iter=200, random_state=42)
modelo_2.fit(X_train_bc_scaled, y_train_bc)
avaliar_modelo_classificacao(modelo_2, X_test_bc_scaled, y_test_bc, "Perceptron")
```

```python
modelo_3 = MLPClassifier(hidden_layer_sizes=(30,), activation='relu', max_iter=150, random_state=42)
modelo_3.fit(X_train_bc_scaled, y_train_bc)
avaliar_modelo_classificacao(modelo_3, X_test_bc_scaled, y_test_bc, "MLP")
```

```python
modelo_4 = MLPClassifier(hidden_layer_sizes=(50, 30), activation='relu', max_iter=1000, random_state=42)
modelo_4.fit(X_train_bc_scaled, y_train_bc)
avaliar_modelo_classificacao(modelo_4, X_test_bc_scaled, y_test_bc, "MLP")
```

```python
modelo_5 = MLPClassifier(hidden_layer_sizes=(30,), activation='tanh', max_iter=1000, random_state=42)
modelo_5.fit(X_train_bc_scaled, y_train_bc)
avaliar_modelo_classificacao(modelo_5, X_test_bc_scaled, y_test_bc, "MLP")
```

```python
modelo_6 = MLPClassifier(hidden_layer_sizes=(30,), activation='relu', max_iter=1000, random_state=42)
modelo_6.fit(X_train_bc, y_train_bc)
avaliar_modelo_classificacao(modelo_6, X_test_bc, y_test_bc, "MLP")
```
