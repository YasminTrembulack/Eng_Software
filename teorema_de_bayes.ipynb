```python
import numpy as np
```

```python
def teorema_de_bayes(p_b_dado_a, p_a, p_b):
    """
    Calcula o Teorema de Bayes: P(A|B) = (P(B|A) * P(A)) / P(B)

    Args:
        p_b_dado_a (float): Probabilidade de B dado A (P(B|A))
        p_a (float): Probabilidade de A (P(A))
        p_b (float): Probabilidade de B (P(B))

    Returns:
        float: Probabilidade de A dado B (P(A|B))
    """
    if p_b == 0:
        raise ValueError("A probabilidade de B não pode ser zero.")
    return (p_b_dado_a * p_a) / p_b

def teorema_de_bayes_com_duas_evidencias(
    p_dado_a1, p_dado_a2, p_a,
    p_dado_b1, p_dado_b2, p_b
):
    numerador = p_dado_a1 * p_dado_a2 * p_a
    denominador = (
        p_dado_a1 * p_dado_a2 * p_a +
        p_dado_b1 * p_dado_b2 * p_b
    )
    return numerador / denominador
```

```python
# Probabilidades do problema
p_doenca = 0.01                  # P(Doente)
p_saudavel = 1 - p_doenca        # P(Saudável)

p_positivo_dado_doente = 0.95    # P(Positivo | Doente)
p_positivo_dado_saudavel = 0.05  # P(Positivo | Saudável)

# Probabilidade de teste positivo: P(Positivo)
p_positivo = (
    p_positivo_dado_doente * p_doenca +
    p_positivo_dado_saudavel * p_saudavel
)

# Aplicando Bayes: P(Doente | Positivo)
p_doente_dado_positivo = teorema_de_bayes(
    p_positivo_dado_doente,
    p_doenca,
    p_positivo
)

print(f"Probabilidade de estar doente dado teste positivo: {p_doente_dado_positivo:.4f} ou {p_doente_dado_positivo*100:.2f}%")
```

```python
# Simulação de pacientes aleatórios
num_pacientes = 1000
resultado_teste = np.random.choice(
    ['positivo', 'negativo'],
    size=num_pacientes,
    p=[p_positivo, 1-p_positivo]
)
```

```python
# Classificação com base na probabilidade de ter doença
linear = p_doente_dado_positivo
classificacoes = [
    'Doente' if np.random.rand() < linear
    else 'Não Doente' for _ in resultado_teste
]

print(f"Probabilidade de estar doente dado teste positivo: {p_doente_dado_positivo*100:.2f}%")
print(f'{'Total de Pacientes: ':<50} {num_pacientes}')
print(f'{'Doentes Classificados: ':<50} {classificacoes.count('Doente')}')
print(f'{'Não Doentes Classificados: ':<50} {classificacoes.count('Não Doente')}')

classificacoes.count('Doente')

```

```python
# Probabilidades do problema
p_animacao = 0.4                    # P(Animação)
p_nao_animacao = 0.6                # P(Não animação)

p_barbie_dado_animacao = 0.8        # P(Barbie | Animação)
p_barbie_dado_nao_animacao = 0.2    # P(Barbie | Não animação)

# Probabilidade de gostar de Barbie: P(Barbie)
p_barbie = (
    p_barbie_dado_animacao * p_animacao +
    p_barbie_dado_nao_animacao * p_nao_animacao
)

# Aplicando Bayes: P(Barbie | Animação)
p_animacao_dado_barbie = teorema_de_bayes(
    p_barbie_dado_animacao,
    p_animacao,
    p_barbie
)

print(f"Probabilidade gostar de Barbie e Animação: {p_animacao_dado_barbie:.4f} ou {p_animacao_dado_barbie*100:.2f}%")
```

1. Qual é a probabilidade total de um jogador apresentar desempenho excepcional?

2. Sabendo que um jogador apresentou desempenho excepcional, qual a probabilidade de ele ser um smurf?

3. Sabendo que um jogador teve desempenho excepcional e também um índice alto de headshot, qual a probabilidade de ele ser um smurf?

# Informações fornecidas:

* A probabilidade de um jogador ser um smurf é 5% (ou seja, 𝑃(𝑆)=0,05).

* Entre os smurfs, 85% tiveram desempenho excepcional: P(D∣S)=0,85
* Entre os não smurfs, 20% tiveram desempenho excepcional: P(D∣S)=0,20

* Entre os smurfs, 70% tiveram um índice alto de headshot: P(H∣S)=0,70
* Entre os não smurfs, 15% tiveram um índice alto de headshot: P(H∣S)=0,15

```python
# Probabilidades do problema
p_smurf = 0.05                        # P(Smurf)
p_nao_smurf = 1 - p_smurf             # P(Não Smurf)

p_d_excepcional_dado_smurf = 0.85     # P(Desempenho Excepcional | Smurf)
p_d_excepcional_dado_nao_smurf = 0.2  # P(Desempenho Excepcional | Não Smurf)

p_headshot_dado_smurf = 0.7           # P(Headshot | Smurf)
p_headshot_dado_nao_smurf = 0.15      # P(Headshot | Não Smurf)

# 1) Qual é a probabilidade total de um jogador apresentar desempenho excepcional?
p_d_excepcional = (
    p_d_excepcional_dado_smurf * p_smurf +
    p_d_excepcional_dado_nao_smurf * p_nao_smurf
)

p_headshot = (
    p_headshot_dado_smurf * p_smurf +
    p_headshot_dado_nao_smurf * p_nao_smurf
)

print(f'Probabilidade de apresentar um desempenho excepcional: {p_d_excepcional*100:.2f}%')

# 2) Sabendo que um jogador apresentou desempenho excepcional, qual a probabilidade de ele ser um smurf?
p_smurf_dado_d_excepcional = teorema_de_bayes(
    p_d_excepcional_dado_smurf,
    p_smurf,
    p_d_excepcional
)


print(f"Probabilidade de apresentar um desempenho excepcional e ser Smurf: {p_smurf_dado_d_excepcional*100:.2f}%")

```

```python
p_smurf_dado_d_e_h = teorema_de_bayes_com_duas_evidencias(
    p_d_excepcional_dado_smurf, p_headshot_dado_smurf, p_smurf,
    
    p_d_excepcional_dado_nao_smurf, p_headshot_dado_nao_smurf, p_nao_smurf
)
print(f"Probabilidade de ser Smurf dado desempenho excepcional e headshot alto: {p_smurf_dado_d_e_h*100:.2f}%")
```
